## 2. Variance theo phut -> chot MIN_MIN=90 + shrinkage K=270
std goals_p90 nhom 0-90 phut = 0.674, gap ~3 lan nhom 270-360 (0.210). Giu MIN_MIN=90 de khong mat dong, dung shrinkage `p90*minutes/(minutes+270)`: hat-trick 90p (3.0) ve 0.75.

```python
bins = pd.read_csv("../data/processed/analytics/eda_minutes_bins.csv", index_col=0)
print(bins.to_string())
print("K=270:", round(3.0 * 90 / (90 + 270), 2))
```
